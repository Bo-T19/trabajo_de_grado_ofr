```python
# =====================================================================
# 3. CONEXIÓN CON BIGQUERY Y FUNCIONES DE APOYO
#    No hay que cambiar nada en esta celda. Se conecta a BigQuery y
#    define las funciones con que el resto del notebook lee y escribe
#    tablas. Es igual en todos los notebooks de esta carpeta.
# =====================================================================
import json
import re
import tempfile
import unicodedata
from datetime import datetime, timezone
from pathlib import Path

import numpy as np
import pandas as pd
from google.cloud import bigquery

NOMBRE_NOTEBOOK = "16_modelos_comparacion"

# --- 3.1 ¿Quedó algo sin llenar en la celda 1? --------------------------
# Si queda algún <<LLENAR ...>>, el notebook se detiene aquí y dice cuál.
_pendientes = sorted(n for n, v in list(globals().items())
                     if n.isupper() and isinstance(v, str) and "<<LLENAR" in v)
if _pendientes:
    raise ValueError("Falta llenar en la celda 1: " + ", ".join(_pendientes))

# --- 3.2 Conexión ---------------------------------------------------------
# En BigQuery el notebook entra con la cuenta de quien lo corre y no
# necesita llave. La llave solo se usa si se corre en un computador.
if globals().get("RUTA_LLAVE_BIGQUERY"):
    from google.oauth2 import service_account
    client = bigquery.Client(
        project=PROYECTO,
        credentials=service_account.Credentials.from_service_account_file(RUTA_LLAVE_BIGQUERY))
else:
    client = bigquery.Client(project=PROYECTO)
print(f"Conectado a BigQuery: proyecto {PROYECTO}")

# Carpeta para los archivos que se descargan mientras corre el notebook.
# Es de paso: todo lo que importa termina en tablas de BigQuery.
_carpeta = globals().get("CARPETA_DESCARGAS") or ""
DIR_TRABAJO = Path(_carpeta) if _carpeta else Path(tempfile.gettempdir()) / "ofr_deforestacion"
DIR_TRABAJO.mkdir(parents=True, exist_ok=True)


# --- 3.3 Nombres de tabla -----------------------------------------------
def nombre_completo(tabla, dataset=None):
    """'panel_ideam' -> 'proyecto.dataset.PANEL_IDEAM'.

    Las tablas que crean estos notebooks se nombran en mayúscula
    (GRILLA, PANEL_IDEAM...). Si la tabla ya viene completa
    (proyecto.dataset.tabla, como las del Observatorio), se devuelve igual.
    Si no, se arma con PROYECTO y el dataset indicado; por defecto,
    DATASET_DEFORESTACION.
    """
    if tabla.count(".") == 2:
        return tabla
    return f"{PROYECTO}.{dataset or globals().get('DATASET_DEFORESTACION')}.{tabla.upper()}"


# --- 3.4 Leer -------------------------------------------------------------
_MARCA = re.compile(r"^(\[índice(?: sin nombre)?\] )?(?:Nombre original: «(.*?)»)?")


_LECTURA_RAPIDA = [True]


def leer_sql(sql):
    """Corre una consulta SQL en BigQuery y devuelve un DataFrame de pandas.

    Usa la lectura rápida de BigQuery (Storage Read API). Si no está
    disponible (por ejemplo, porque la cuenta no tiene el rol «BigQuery
    Read Session User»), sigue con la lectura normal, que da el mismo
    resultado pero tarda más en tablas grandes.
    """
    trabajo = client.query(sql)
    trabajo.result()                      # si la consulta tiene un error, sale aquí
    resultado = None
    if _LECTURA_RAPIDA[0]:
        try:
            resultado = trabajo.to_arrow()
        except Exception as e:
            _LECTURA_RAPIDA[0] = False
            print("Aviso: la lectura rápida de BigQuery no está disponible "
                  f"({type(e).__name__}); se usa la lectura normal.")
    if resultado is None:
        resultado = trabajo.to_arrow(create_bqstorage_client=False)
    return resultado.to_pandas(date_as_object=False, coerce_temporal_nanoseconds=True)


def leer_tabla(tabla, columnas=None, dataset=None, filtro=None):
    f"""Lee una tabla de BigQuery a pandas.

    columnas  lista de columnas a traer; por defecto, todas.
    filtro    condición SQL opcional, p. ej. "anio >= {ANIO_INICIO}".

    Si la tabla se guardó con guardar_tabla(), devuelve las filas en el
    mismo orden en que se guardaron, las columnas con su nombre original
    (tildes, espacios) y el índice.
    """
    destino = nombre_completo(tabla, dataset)
    esquema = client.get_table(destino).schema
    con_orden = any(c.name == "orden_fila" for c in esquema)
    seleccion = ", ".join(f"`{c}`" for c in columnas) if columnas else "*"
    df = leer_sql(f"SELECT {seleccion} FROM `{destino}`"
                  + (f" WHERE {filtro}" if filtro else "")
                  + (" ORDER BY orden_fila" if con_orden else ""))
    df = df.drop(columns=["orden_fila"], errors="ignore")

    originales, indice, sin_nombre = {}, [], set()
    for campo in esquema:
        marca = _MARCA.match(campo.description or "")
        if marca.group(2):
            originales[campo.name] = marca.group(2)
        if marca.group(1) and campo.name in df.columns:
            indice.append(campo.name)
            if "sin nombre" in marca.group(1):
                sin_nombre.add(campo.name)
    if indice:
        df = df.set_index(indice)
        df.index = df.index.set_names(
            [None if n in sin_nombre else originales.get(n, n) for n in indice])
    return df.rename(columns=originales)


# --- 3.5 Escribir -----------------------------------------------------------
_DATASETS_LISTOS = set()


def _nombre_para_bigquery(nombre, usados):
    """Nombre de columna que BigQuery acepta siempre: letras sin tilde,
    números y guion bajo. Los que ya cumplen se dejan igual."""
    texto = str(nombre)
    if re.fullmatch(r"[A-Za-z_][A-Za-z0-9_]{0,299}", texto) and texto.lower() not in usados:
        usados.add(texto.lower())
        return texto
    limpio = unicodedata.normalize("NFKD", texto).encode("ascii", "ignore").decode()
    limpio = re.sub(r"[^0-9A-Za-z]+", "_", limpio).strip("_").lower() or "columna"
    if limpio[0].isdigit():
        limpio = "c_" + limpio
    limpio, base, k = limpio[:280], limpio[:280], 2
    while limpio.lower() in usados:
        limpio, k = f"{base}_{k}", k + 1
    usados.add(limpio.lower())
    return limpio


def _tipos_para_bigquery(df):
    """Deja cada columna en un tipo que BigQuery reciba sin error:
    categorías a texto, enteros con vacíos a entero, mezclas a texto."""
    for c in df.columns:
        s = df[c]
        if isinstance(s.dtype, pd.CategoricalDtype):
            s = s.astype(object).where(s.notna(), None)
        if s.dtype == object:
            valores = s.dropna()
            tipos = {type(v) for v in valores}
            if not tipos or tipos <= {str}:
                pass
            elif all(isinstance(v, (bool, np.bool_)) for v in valores):
                s = s.astype("boolean")
            elif all(isinstance(v, (int, np.integer)) and not isinstance(v, bool) for v in valores):
                s = s.astype("Int64")
            elif all(isinstance(v, (int, float, np.integer, np.floating)) for v in valores):
                s = s.astype(float)
            else:
                s = s.map(lambda v: None if v is None or v is pd.NA
                          or (isinstance(v, float) and v != v) else str(v))
        df[c] = s
    return df


def guardar_tabla(df, tabla, descripcion, dataset=None, indice=False,
                  descripciones=None, procedencia=None):
    """Escribe `df` en BigQuery. Si la tabla ya existe, la REEMPLAZA entera.

    tabla          nombre corto, p. ej. "panel_ideam".
    descripcion    texto que se ve en BigQuery, en los detalles de la tabla.
    dataset        por defecto, DATASET_DEFORESTACION. Si no existe, se crea.
    indice         True para guardar también el índice de pandas como
                   columna(s); leer_tabla() lo recupera.
    descripciones  {columna: texto}, para documentar columnas en BigQuery.
    procedencia    {"fuente": ..., ...}: si se da, deja una fila en la tabla
                   registro_cargas con qué se cargó, de dónde y cuándo.
    """
    dataset = dataset or globals().get("DATASET_DEFORESTACION")
    destino = nombre_completo(tabla, dataset)
    if dataset not in _DATASETS_LISTOS:
        ds = bigquery.Dataset(f"{PROYECTO}.{dataset}")
        ds.location = globals().get("UBICACION", "US")
        client.create_dataset(ds, exists_ok=True)
        _DATASETS_LISTOS.add(dataset)

    datos = df.copy()
    marcas = {}
    if indice:
        nombres_indice = list(datos.index.names)
        datos = datos.reset_index()
        for k, n in enumerate(nombres_indice):
            columna = datos.columns[k]
            marcas[columna] = "[índice sin nombre] " if n is None else "[índice] "
    else:
        datos = datos.reset_index(drop=True)
    if isinstance(datos.columns, pd.MultiIndex):
        datos.columns = [" | ".join(str(x) for x in c if str(x) != "") for c in datos.columns]

    descripciones = descripciones or {}
    usados, notas, columnas = set(), {}, []
    for c in datos.columns:
        nuevo = _nombre_para_bigquery(c, usados)
        texto = marcas.get(c, "")
        if nuevo != str(c):
            texto += f"Nombre original: «{c}»"
        if descripciones.get(c):
            texto += (". " if texto and not texto.endswith("] ") else "") + descripciones[c]
        notas[nuevo] = texto[:1024]
        columnas.append(nuevo)
    datos.columns = columnas
    datos = _tipos_para_bigquery(datos)
    # BigQuery no guarda el orden de las filas, y algunos cálculos (el
    # muestreo del random forest, por ejemplo) dependen de él. Esta
    # columna lo conserva; leer_tabla() la usa y la quita.
    datos["orden_fila"] = np.arange(len(datos), dtype="int64")
    notas["orden_fila"] = ("Posición de la fila en la tabla original. Sirve para leerla "
                           "en el mismo orden: ORDER BY orden_fila.")

    trabajo = client.load_table_from_dataframe(
        datos, destino,
        job_config=bigquery.LoadJobConfig(write_disposition="WRITE_TRUNCATE"))
    trabajo.result()

    t = client.get_table(destino)
    t.description = descripcion
    t.schema = [bigquery.SchemaField(f.name, f.field_type, mode=f.mode,
                                     description=notas.get(f.name) or None,
                                     fields=f.fields) for f in t.schema]
    client.update_table(t, ["description", "schema"])
    print(f"Guardada {destino}: {t.num_rows:,} filas, {len(t.schema)} columnas")

    if procedencia is not None:
        fila = pd.DataFrame([{
            "tabla": tabla.upper(),
            "filas": int(t.num_rows),
            "fuente": str(procedencia.get("fuente", "")),
            "parametros": json.dumps({k: v for k, v in procedencia.items() if k != "fuente"},
                                     ensure_ascii=False, default=str),
            "notebook": NOMBRE_NOTEBOOK,
            "cargado_en": datetime.now(timezone.utc),
        }])
        client.load_table_from_dataframe(
            fila, nombre_completo("registro_cargas", dataset),
            job_config=bigquery.LoadJobConfig(write_disposition="WRITE_APPEND")).result()
```

```python
def etiqueta(m):
    for k, v in [("precisión promedio", "AP"), ("R² excluyendo", "R² sin Pedraza"), ("R² una sola", "R² una sola regresión (sin separar)"),
                 ("R² en escala log", "R² escala log"), ("correlación de Spearman", "Spearman"), ("error absoluto promedio", "MAE"),
                 ("error absoluto mediano", "MedAE")]:
        if m.startswith(k):
            return v
    return m


def leer_metricas(n):
    d = leer_resultado(f"{MODELOS[n]['prefijo']}_metricas")
    d["parte"] = np.where(d.modelo.str.contains("deforesta"), "¿deforesta?", "¿cuánto?")
    d["métrica"] = d["métrica"].map(etiqueta)
    return d


MET = {n: leer_metricas(n) for n in NOMS}


def tabla_met(parte, columna):
    t = pd.DataFrame({n: d[d.parte == parte].set_index("métrica")[columna] for n, d in MET.items()})
    return t


def metricas_prob(y, prob, umbral=UMBRAL):
    pred = (prob >= umbral).astype(int)
    tn, fp, fn, tp = confusion_matrix(y, pred).ravel()
    recall, espec = tp / (tp + fn), tn / (tn + fp)
    return {"exactitud balanceada": (recall + espec) / 2, "AUC": roc_auc_score(y, prob), "precisión": tp / (tp + fp),
            "recall / sensibilidad": recall, "especificidad": espec, "F1": f1_score(y, pred),
            "AP": average_precision_score(y, prob), "tn": tn, "fp": fp, "fn": fn, "tp": tp}


REC = pd.DataFrame({n: metricas_prob(y25, P[f"prob|{n}"].values) for n in NOMS})

# Verificacion: lo recalculado aqui vs. lo que cada notebook reporto en su propio CSV (prueba 2025)
rep = tabla_met("¿deforesta?", "prueba_temporal_2025")
dif = (REC.loc[rep.index.intersection(REC.index)] - rep.loc[rep.index.intersection(REC.index)]).abs()
for n in NOMS:
    verificaciones.append((f"{n}: lo recalculado aquí (AUC, exactitud balanceada, precisión, recall, F1...) coincide con su CSV "
                           f"(diferencia máxima {dif[n].max():.4f})", bool(dif[n].max() < 5e-3)))
for texto, ok in verificaciones:
    print(("OK       " if ok else "REVISAR  ") + texto)
if not all(ok for _, ok in verificaciones):
    print("\nHay verificaciones en REVISAR: lo más probable es que alguno de los notebooks de modelo se haya corrido con datos o semilla distintos. "
          "Corra los tres de nuevo, de principio a fin, y vuelva a correr este.")
```

## 3. Curvas ROC, precisión-recall, ganancia y calibración (prueba 2025)

Las cuatro lecturas de la misma probabilidad, superpuestas. **ROC** y **precisión-recall** miden qué tan bien ordena el modelo; la **curva de
ganancia** responde la pregunta operativa del Observatorio: *si reviso solo el 10, 20 o 30 % de los municipios con mayor riesgo, ¿qué proporción de lo que
deforesta cubro?* (OBJ-N1); y la **calibración** dice si la probabilidad se puede leer como un número (un 0,8 que de verdad ocurre 8 de cada 10 veces) o solo como un orden.

```python
fig, ax = plt.subplots(1, 4, figsize=(22, 5))
ha25 = P.def_ideam_ha.fillna(0).values
orden_ganancia = {}
for n in NOMS:
    prob = P[f"prob|{n}"].values
    fpr, tpr, _ = roc_curve(y25, prob)
    ax[0].plot(fpr, tpr, color=COLOR[n], label=f"{n} (AUC = {REC.loc['AUC', n]:.3f})")
    pr_, rc_, _ = precision_recall_curve(y25, prob)
    ax[1].plot(rc_, pr_, color=COLOR[n], label=f"{n} (AP = {REC.loc['AP', n]:.3f})")
    o = np.argsort(-prob, kind="stable")
    orden_ganancia[n] = o
    x = np.arange(1, len(o) + 1) / len(o)
    ax[2].plot(x, np.cumsum(y25[o]) / y25.sum(), color=COLOR[n], label=n)
    frac_pos, frac_pred = calibration_curve(y25, prob, n_bins=8, strategy="quantile")
    ax[3].plot(frac_pred, frac_pos, "o-", color=COLOR[n], label=f"{n} (Brier = {brier_score_loss(y25, prob):.3f})")
ax[0].plot([0, 1], [0, 1], "--", color="grey", lw=1)
ax[0].set(xlabel="falsos positivos", ylabel="verdaderos positivos", title=f"ROC — prueba {ANIO_PRUEBA}")
ax[1].set(xlabel="recall", ylabel="precisión", title=f"Precisión-recall — prueba {ANIO_PRUEBA}")
ax[2].plot([0, 1], [0, 1], "--", color="grey", lw=1, label="al azar")
ax[2].set(xlabel="% de municipios revisados (mayor riesgo primero)", ylabel="% de los que deforestan cubiertos", title="Curva de ganancia (municipios)")
ax[3].plot([0, 1], [0, 1], "--", color="grey", lw=1)
ax[3].set(xlabel="probabilidad predicha", ylabel="frecuencia observada", title="Calibración")
for a in ax:
    a.legend(fontsize=8)
plt.tight_layout()
plt.show()

filas = []
for n in NOMS:
    o = orden_ganancia[n]
    for pct in (10, 20, 30):
        k = int(np.ceil(len(o) * pct / 100))
        filas.append({"modelo": n, "se revisa el": f"{pct} %", "municipios que deforestan cubiertos (%)": 100 * y25[o[:k]].sum() / y25.sum(),
                      "hectáreas deforestadas cubiertas (%)": 100 * ha25[o[:k]].sum() / ha25.sum()})
cobertura = pd.DataFrame(filas).pivot(index="se revisa el", columns="modelo", values=["municipios que deforestan cubiertos (%)", "hectáreas deforestadas cubiertas (%)"])
display(cobertura.round(1))
brier = pd.Series({n: brier_score_loss(y25, P[f"prob|{n}"]) for n in NOMS}, name="Brier (menor = mejor calibrada)")
display(brier.round(3).to_frame().T)
```

### 4.1 ¿Dónde se equivoca cada modelo? Falsos positivos y falsos negativos

Un **falso negativo** (municipio que deforestó y no se marcó) deja pasar deforestación; un **falso positivo** (alarma en uno que no deforestó) gasta una visita de
verificación. Aquí se ve (a) en qué tamaño de bosque ocurren, (b) cuántas hectáreas dejan pasar los falsos negativos y (c) si los tres se equivocan en los mismos municipios.
El tamaño se mide en **hectáreas de bosque** (% de bosque × área): es la variable que, sola, ya ordena bien el riesgo (sección 7.1b).

```python
P["ha_bosque"] = P.bosque * P.area  # % de bosque x km2 = hectareas de bosque
P["cuartil_bosque"] = pd.qcut(P.ha_bosque, 4, labels=["Q1 (menos bosque)", "Q2", "Q3", "Q4 (más bosque)"])
PRED = {n: (P[f"prob|{n}"].values >= UMBRAL).astype(int) for n in NOMS}

filas = []
for n in NOMS:
    fp, fn = (PRED[n] == 1) & (y25 == 0), (PRED[n] == 0) & (y25 == 1)
    for q in P.cuartil_bosque.cat.categories:
        m = (P.cuartil_bosque == q).values
        filas.append({"modelo": n, "cuartil de bosque": q, "municipios": int(m.sum()), "deforestan": int(y25[m].sum()),
                      "falsos positivos": int((fp & m).sum()), "falsos negativos": int((fn & m).sum())})
err_cuartil = pd.DataFrame(filas)
display(err_cuartil.pivot(index="cuartil de bosque", columns="modelo", values=["falsos negativos", "falsos positivos"]).reindex(columns=NOMS, level=1))

ha_tot = ha25.sum()
fn_ha = pd.DataFrame({n: {"falsos negativos": int(((PRED[n] == 0) & (y25 == 1)).sum()),
                          "hectáreas que dejan pasar": ha25[(PRED[n] == 0) & (y25 == 1)].sum(),
                          "% de las hectáreas deforestadas": 100 * ha25[(PRED[n] == 0) & (y25 == 1)].sum() / ha_tot,
                          "falsos positivos": int(((PRED[n] == 1) & (y25 == 0)).sum())} for n in NOMS})
display(fn_ha.round(1))

err = pd.DataFrame({n: PRED[n] != y25 for n in NOMS})
en_todos, en_alguno_e = int(err.all(axis=1).sum()), int(err.any(axis=1).sum())
print(f"Municipios mal clasificados por los TRES modelos: {en_todos}; por al menos uno: {en_alguno_e}. "
      f"Errores por modelo: {', '.join(f'{n} {int(err[n].sum())}' for n in NOMS)}.")
print("Los falsos negativos se concentran donde hay poco bosque (un municipio de 100 ha de bosque que deforesta 0,2 ha cuenta igual que uno de miles); "
      "los falsos positivos, en tamaños intermedios. Ver la tabla por cuartil.")
```

**Lectura: dónde se equivocan (falsos positivos y negativos).**

- **Los errores son pocos y casi siempre en municipios pequeños.** Con umbral 0,5, los falsos negativos son 65, 68 y 82 y los falsos positivos 29, 29 y 34 (gradient boosting, bosque, regresión logística). Los falsos negativos están casi todos en los dos cuartiles con menos bosque (55 de 65, 60 de 68 y 81 de 82) y en el cuartil de más bosque **no hay ningún error**.
- **Lo que se deja pasar pesa poco en hectáreas.** Los falsos negativos suman unas 384, 371 y 896 hectáreas, es decir 0,3 %, 0,3 % y 0,8 % de las deforestadas. La alarma que se pierde es sobre todo la de municipios con casi nada de bosque, donde "deforesta > 0" se cruza por muy poco. Los falsos positivos son visitas de verificación (29 a 34) y se
  concentran en tamaños intermedios.
- **Los tres se equivocan en los mismos municipios.** 72 municipios son mal clasificados por los tres y 139 por al menos uno; 72 de los 94 errores del gradient boosting (77 %) son compartidos. Es decir, la dificultad está en los datos (municipios pequeños, deforestación cercana a cero) más que en la técnica.

### 4.2 Misma regla de marcado: el mismo número de municipios para los tres, y distintos umbrales

Con umbral 0,5 cada modelo marca una cantidad distinta de municipios, y eso por sí solo cambia sus falsos positivos y negativos. Para comparar **sin ese efecto**, se le pide a cada modelo
que marque los *K* municipios de mayor probabilidad, con *K* = los que realmente deforestaron en 2025 (así, falsos positivos = falsos negativos). También se muestra qué pasa al mover el umbral.

```python
K = int(y25.sum())
filas = []
for n in NOMS:
    o = np.argsort(-P[f"prob|{n}"].values, kind="stable")[:K]
    tp = int(y25[o].sum())
    filas.append({"modelo": n, "marcados con umbral 0,5": int(PRED[n].sum()), f"aciertos entre los {K} de mayor probabilidad": tp,
                  "falsos negativos (= falsos positivos)": K - tp, "precisión = recall": tp / K})
mismo_k = pd.DataFrame(filas).set_index("modelo")
display(mismo_k.round(3))

filas = []
for n in NOMS:
    for u in (0.3, 0.5, 0.7):
        pr_ = (P[f"prob|{n}"].values >= u).astype(int)
        tn, fp_, fn_, tp = confusion_matrix(y25, pr_).ravel()
        filas.append({"modelo": n, "umbral": u, "marcados": int(pr_.sum()), "falsos negativos": int(fn_), "falsos positivos": int(fp_),
                      "recall": tp / (tp + fn_), "especificidad": tn / (tn + fp_)})
umbrales = pd.DataFrame(filas)
display(umbrales.pivot(index="umbral", columns="modelo", values=["falsos negativos", "falsos positivos"]).reindex(columns=NOMS, level=1))
dif_k = mismo_k.iloc[:, 2]
print(f"Con la misma regla (marcar {K}), los falsos negativos van de {int(dif_k.min())} a {int(dif_k.max())}; con umbral 0,5 los marcados van de "
      f"{int(mismo_k.iloc[:, 0].min())} a {int(mismo_k.iloc[:, 0].max())}. Una parte de las diferencias de sensibilidad del umbral 0,5 viene del punto de operación, no de que un modelo ordene mejor.")
```

**Lectura: parte de las diferencias de sensibilidad es el punto de operación.**

Si cada modelo marca los 410 municipios de mayor probabilidad (los que de verdad deforestaron), los falsos negativos son iguales a los falsos positivos: 47, 47 y 56, y precisión = recall de 0,885, 0,885 y 0,863. El gradient boosting y el bosque quedan **idénticos**, y la regresión logística, algo por debajo. Con umbral 0,5, en cambio, marcan 374, 371 y 362 municipios.
Esto desarma el argumento de que un modelo "detecta más" por tener mayor sensibilidad: gran parte viene del umbral. El umbral mismo es una decisión de negocio: bajarlo a 0,3 reduce los falsos negativos (39, 32, 41) a cambio de más falsas alarmas (62, 66, 80); subirlo a 0,7 hace lo contrario (115, 118, 165 negativos y 13, 13, 7 positivos).

### 4.3 Lo mismo, en términos de *odds*

El *odds* es la razón entre los que deforestan y los que no: **deforestan / no deforestan**. Un *odds* de 1 equivale a "1 de cada 2" (50 %); un *odds* de 2, a dos que deforestan por cada uno que no (67 %). Sin ningún modelo, en 2025 deforestan 410 de 639 municipios, así que el *odds* de partida es 410/229 = 1,79.

Cuatro formas de leer lo que aporta cada modelo, con la misma regla para los tres:

1. **Razón de *odds* diagnóstica.** Compara el *odds* de deforestar entre los municipios que el modelo marca y entre los que no marca: (aciertos de deforestación × aciertos de no deforestación) / (falsos positivos × falsos negativos). Resume la matriz de confusión en un solo número.
2. **Razón de verosimilitud positiva (LR+).** Cuántas veces sube el *odds* de deforestar cuando el modelo marca un municipio: *odds* de partida × LR+ = *odds* entre los marcados. La LR− hace lo mismo con los no marcados (menor que 1 es mejor).
3. ***Odds* por tramo de riesgo.** El *odds* observado en el 30 % de mayor riesgo de cada modelo, en el siguiente 30 % y en el 40 % más bajo. En el tramo alto casi no quedan municipios que no deforesten, así que su *odds* es muy sensible a un solo caso: se lee como "casi todos deforestan" y no como una cifra exacta; por eso se muestran también los conteos.
4. ***Odds* por tamaño de bosque**, sin modelo. Muestra cuánto del resultado viene solo del tamaño del municipio (el sesgo estructural de la sección 7.1b), y qué *odds* predice cada modelo en cada cuartil.

Las dos primeras se muestran con dos reglas de marcado: el umbral 0,5 y los 410 municipios de mayor probabilidad (el mismo número para los tres, como en 4.2). Si un denominador es cero, el *odds* queda vacío (no se inventa un valor).

```python
def odds(a, b):
    return a / b if b > 0 else np.nan


n_pos = int(y25.sum())
n_neg = int(len(y25) - n_pos)
odds_base = odds(n_pos, n_neg)
print(f"Sin ningún modelo: deforestan {n_pos} y no deforestan {n_neg}; odds = {n_pos}/{n_neg} = {odds_base:.2f} (probabilidad {n_pos / len(y25):.1%}).")


def resumen_odds(pred, nombre, regla):
    tn, fp_, fn_, tp = confusion_matrix(y25, pred).ravel()
    recall, espec = tp / (tp + fn_), tn / (tn + fp_)
    return {"modelo": nombre, "regla de marcado": regla, "marcados": int(pred.sum()),
            "odds entre los marcados": odds(tp, fp_), "odds entre los no marcados": odds(fn_, tn),
            "razón de odds diagnóstica": odds(tp * tn, fp_ * fn_),
            "LR+": odds(recall, 1 - espec), "LR−": odds(1 - recall, espec)}


filas = []
for n in NOMS:
    filas.append(resumen_odds((P[f"prob|{n}"].values >= UMBRAL).astype(int), n, "umbral 0,5"))
    o_k = np.argsort(-P[f"prob|{n}"].values, kind="stable")[:n_pos]
    pred_k = np.zeros(len(P), dtype=int)
    pred_k[o_k] = 1
    filas.append(resumen_odds(pred_k, n, f"los {n_pos} de mayor probabilidad"))
odds_modelos = pd.DataFrame(filas)
display(odds_modelos.round(2))

u05 = odds_modelos[odds_modelos["regla de marcado"] == "umbral 0,5"].set_index("modelo")
print("Comprobación (odds de partida × LR+ = odds entre los marcados):",
      "OK" if np.allclose(odds_base * u05["LR+"], u05["odds entre los marcados"]) else "NO COINCIDE")

# Odds por tramo de riesgo de cada modelo
TRAMOS = [("30 % de mayor riesgo", 0.0, 0.30), ("siguiente 30 %", 0.30, 0.60), ("40 % de menor riesgo", 0.60, 1.0)]
filas = []
for n in NOMS:
    orden_n = np.argsort(-P[f"prob|{n}"].values, kind="stable")
    for nombre, a, b in TRAMOS:
        idx = orden_n[int(round(a * len(P))):int(round(b * len(P)))]
        d = int(y25[idx].sum())
        filas.append({"modelo": n, "tramo de riesgo": nombre, "municipios": len(idx), "deforestan": d,
                      "no deforestan": len(idx) - d, "odds": odds(d, len(idx) - d)})
odds_tramos = pd.DataFrame(filas)
display(odds_tramos.pivot(index="tramo de riesgo", columns="modelo", values="odds").reindex([t[0] for t in TRAMOS])[NOMS].round(2))
display(odds_tramos.round(2))

# Odds por cuartil de bosque, observado y predicho
filas = []
for q in P.cuartil_bosque.cat.categories:
    m = (P.cuartil_bosque == q).values
    d = int(y25[m].sum())
    fila = {"cuartil de bosque": q, "municipios": int(m.sum()), "deforestan": d, "no deforestan": int(m.sum()) - d, "odds observado": odds(d, int(m.sum()) - d)}
    for n in NOMS:
        pm = float(P.loc[m, f"prob|{n}"].mean())
        fila[f"odds predicho ({n})"] = pm / (1 - pm)
    filas.append(fila)
odds_cuartil = pd.DataFrame(filas)
display(odds_cuartil.round(2))
q1, q4 = odds_cuartil["odds observado"].iloc[0], odds_cuartil["odds observado"].iloc[-1]
print(f"El odds de deforestar en el cuartil con más bosque es {q4 / q1:.1f} veces el del cuartil con menos bosque, sin usar ningún modelo." if q1 > 0 and np.isfinite(q4) else "No se puede comparar los cuartiles extremos (un denominador es cero).")

fig, ax = plt.subplots(1, 2, figsize=(13, 4.3))
anchos = 0.26
for k, n in enumerate(NOMS):
    o = odds_tramos[odds_tramos.modelo == n].set_index("tramo de riesgo").odds.reindex([t[0] for t in TRAMOS])
    ax[0].bar(np.arange(3) + (k - 1) * anchos, o.values, anchos, color=COLOR[n], label=n)
ax[0].axhline(odds_base, color="gray", ls="--", lw=1)
ax[0].text(2.45, odds_base, f"odds de partida {odds_base:.2f}", ha="right", va="bottom", fontsize=8, color="gray")
ax[0].set_xticks(range(3), [t[0] for t in TRAMOS]); ax[0].set_yscale("log"); ax[0].set_ylabel("odds de deforestar (escala log)")
ax[0].set_title("Odds por tramo de riesgo de cada modelo"); ax[0].legend(frameon=False, fontsize=8)
ax[1].bar(np.arange(4), odds_cuartil["odds observado"], color="#8c8c8c")
ax[1].axhline(odds_base, color="gray", ls="--", lw=1)
ax[1].set_xticks(range(4), odds_cuartil["cuartil de bosque"], fontsize=8); ax[1].set_yscale("log")
ax[1].set_title("Odds observado por cuartil de bosque (sin modelo)")
plt.tight_layout()
plt.show()
```

**Lectura: los tres modelos en términos de *odds* (lo que pidió el tutor).**

- **Qué aporta marcar un municipio.** Sin modelo, el *odds* de deforestar es 1,79. Con el umbral 0,5, entre los municipios que el modelo marca el *odds* sube a 11,9 (gradient boosting), 11,8 (bosque) y 9,7 (regresión logística): unos 12, 12 y 10 que deforestan por cada uno que no. Entre los que no marca baja a 0,32, 0,34 y 0,42. Dicho con la razón de verosimilitud, marcar multiplica el *odds* de partida por 6,6, 6,6 y 5,4 (LR+), y no marcar lo deja en 0,18, 0,19 y 0,23 veces el de partida (LR−).
- **La razón de *odds* diagnóstica resume si los modelos son iguales o no.** Con la misma regla de marcado (los 410 de mayor probabilidad) vale 29,9 para el gradient boosting y el bosque, idénticos, y 19,5 para la regresión logística, cerca de dos tercios. Con el umbral 0,5 vale 36,6, 34,7 y 22,9. Es la misma conclusión de la comparación: gradient boosting y bosque son equivalentes y la regresión logística queda algo por debajo en 2025. No hay una prueba estadística de la diferencia; es una lectura de la tabla.
- **Por tramo de riesgo el orden es el esperado en los tres.** En el segundo 30 % el *odds* es 4,97, 5,37 y 3,34, y en el 40 % de menor riesgo 0,31, 0,30 y 0,39. El tramo alto tiene 191, 190 y 191 municipios que deforestan de 192 (uno o dos que no), así que su *odds* (191, 95 y 191) depende de un solo caso: se lee como "casi todos deforestan", no como una cifra exacta ni para comparar modelos. Incluso en el 40 % de menor riesgo deforestan 60, 59 y 72 de 256 municipios: los modelos ordenan bien, pero no descartan.
- **El tamaño del bosque, solo, ya cambia mucho el *odds*.** Sin ningún modelo, el *odds* observado es 0,31 en el cuartil con menos bosque (38 de 160 deforestan), 0,98 en el segundo y 5,12 en el tercero, unas 16 veces más que en el primero. En el cuartil con más bosque deforestan los 160: no hay ni un municipio que no, y el *odds* no se puede calcular. Es la cifra del sesgo estructural de la sección 7.1b, y por eso allí el AUC se mide dentro de los tres primeros cuartiles. Los modelos siguen el mismo patrón: 0,15 a 0,18 en el primer cuartil y 2,6 a 3,8 en el tercero.
- **Los tres esperan menos deforestación de la que hubo.** En cada cuartil el *odds* predicho (el de la probabilidad media del cuartil) queda por debajo del observado: por ejemplo 0,15 a 0,18 contra 0,31 en el primero y 2,6 a 3,8 contra 5,12 en el tercero. La calibración global ya se mostró con el Brier de la sección 3; aquí solo se anota.
- **Cómo usarlo.** La razón de *odds* diagnóstica y la LR+, con las dos reglas de marcado, sustentan que los modelos son parecidos en lo esencial. Los tramos y el cuartil de bosque dan la lectura de negocio y muestran el sesgo. Son asociaciones, no causas.

## 7. ¿Qué variables importan? — la comparación del eje central de ODM-2 y ODM-3

La medida común es la **importancia por permutación sobre 2025** (cuánto cae el AUC al barajar una variable), que los tres notebooks calculan con el mismo código
y guardan con el mismo esquema. Se separan las dos clases de variables: las **estructurales** (bosque y área del municipio, que ninguna política "mueve") y las **13
de intervención** (crédito, cultivos y ganado). Las magnitudes son comparables en orden y proporción, no como números absolutos entre técnicas
(una variable correlacionada con otra reparte su importancia de forma distinta en un árbol que en una regresión).

```python
etiqueta_a_clave = {v: k for k, v in NOMBRES.items()}


def leer_perm(n):
    d = leer_resultado(f"{MODELOS[n]['prefijo']}_importancia_permutacion_clasificacion")
    d["clave"] = d["variable"].map(etiqueta_a_clave)
    if d.clave.isna().any():
        raise RuntimeError(f"{n}: variables sin correspondencia con las 15 de gradient boosting: {d.loc[d.clave.isna(), 'variable'].tolist()}")
    return d.set_index("clave")


PERM = {n: leer_perm(n) for n in NOMS}
imp = pd.DataFrame({n: PERM[n]["importancia_media"] for n in NOMS}).reindex(EXPL)
excluidas_vif = [k for k in PERM["Regresión logística"].index if str(PERM["Regresión logística"].loc[k].get("tipo", "")).startswith("excluida")]

# 7.1 estructural vs. intervencion
imp_pos = imp.clip(lower=0)
part_estructural = 100 * imp_pos.loc[ESTRUCTURALES].sum() / imp_pos.sum()
print("Participación de las variables ESTRUCTURALES (bosque y área) en la importancia total, por modelo:")
print("  " + "; ".join(f"{n} {part_estructural[n]:.0f} %" for n in NOMS))
espec = {n: leer_resultado(f"{MODELOS[n]['prefijo']}_especificaciones")["AUC"] for n in NOMS}
for n, s in espec.items():
    a = s[[i for i in s.index if i.startswith("A.")][0]]
    b = s[[i for i in s.index if i.startswith("B.")][0]]
    print(f"  {n}: solo con bosque y área (especificación A) el AUC es {a:.3f}; con el modelo completo (B) {b:.3f} → "
          f"el crédito, los cultivos y el ganado aportan {b - a:+.3f} de AUC.")
```

### 7.1b ¿Hay un sesgo estructural? Referencias de tamaño y lectura dentro de cada tamaño de bosque

Que bosque y área dominen no es un error de código: la variable objetivo es "deforesta" (alguna pérdida de bosque en el año, tasa > 0), y un municipio con mucho bosque tiene mecánicamente más
probabilidad de perder alguna parte. Para medirlo: (1) cuánto AUC logra una *sola* variable de tamaño, sin modelo; (2) cuánto logra cada modelo **dentro** de un mismo
cuartil de bosque (donde el tamaño ya no ayuda tanto); (3) el reparto de la importancia entre las 15 variables, medida como caída del AUC al barajar la variable
(la figura de las 13 variables de intervención normaliza solo entre ellas, por eso no se ve este peso).

```python
def auc_seguro(y, s):
    y = np.asarray(y); return roc_auc_score(y, s) if 0 < y.sum() < len(y) else np.nan

ref = {"solo hectáreas de bosque": P.ha_bosque.values, "solo área del municipio": P.area.values, "solo bosque (% del municipio)": P.bosque.values}
cuart = P.cuartil_bosque.astype(str).values
q1_3 = np.isin(cuart, [c for c in P.cuartil_bosque.cat.categories[:3]])
filas = []
for nombre, s in list(ref.items()) + [(n, P[f"prob|{n}"].values) for n in NOMS]:
    fila = {"referencia / modelo": nombre, "AUC global": auc_seguro(y25, s)}
    for q in P.cuartil_bosque.cat.categories[:3]:
        m = cuart == str(q)
        fila[f"AUC dentro de {q.split(' ')[0]}"] = auc_seguro(y25[m], s[m])
    fila["AUC en Q1-Q3 juntos"] = auc_seguro(y25[q1_3], s[q1_3])
    filas.append(fila)
ref_tamano = pd.DataFrame(filas).set_index("referencia / modelo")
display(ref_tamano.round(3))
q4 = (cuart == str(P.cuartil_bosque.cat.categories[3]))
print(f"% de municipios que deforestan por cuartil de bosque: " + "; ".join(f"{str(q).split(' ')[0]} {100 * y25[cuart == str(q)].mean():.0f} %" for q in P.cuartil_bosque.cat.categories)
      + f". En Q4 casi todos deforestan ({int(y25[q4].sum())} de {int(q4.sum())}), por eso ahí el AUC no se puede calcular o no informa.")
mejor_ref = ref_tamano.iloc[:3]["AUC global"].max()
print(f"Una sola variable de tamaño logra un AUC de {mejor_ref:.3f}; los modelos, {ref_tamano.loc[NOMS, 'AUC global'].min():.3f} a {ref_tamano.loc[NOMS, 'AUC global'].max():.3f}. "
      f"Dentro de Q1-Q3 los modelos logran {ref_tamano.loc[NOMS, 'AUC en Q1-Q3 juntos'].min():.3f} a {ref_tamano.loc[NOMS, 'AUC en Q1-Q3 juntos'].max():.3f} frente a "
      f"{ref_tamano.iloc[:3]['AUC en Q1-Q3 juntos'].max():.3f} de la mejor referencia de tamaño: lo que agregan crédito, cultivos y ganado se nota ahí, no en el AUC global.")
```

### 7.4 Dirección de la asociación (solo asociación, no causalidad)

Gradient boosting y random forest no entregan un signo por variable como una regresión; para poder comparar los tres con la misma vara se calcula, para cada
modelo, la **correlación de Spearman entre el valor de la variable y el riesgo que ese modelo predijo en 2025**. Es una lectura marginal (no controla por las demás variables): sirve para ver si
los tres coinciden en el sentido, no como efecto de cada variable. La última columna es el *odds ratio* de la regresión logística (controlado por las demás y con errores agrupados por departamento): menor que 1,
la probabilidad baja cuando la variable sube.

```python
rho_dir = pd.DataFrame({n: {v: stats.spearmanr(P[v], P[f"prob|{n}"])[0] for v in INTERV} for n in NOMS})
coefs = leer_resultado("regresion_logistica_coeficientes_inferencia")
dir_tabla = rho_dir.copy()
dir_tabla["OR de la logística (por +1 desv.)"] = [coefs["OR"].get(NOMBRES[v], np.nan) for v in INTERV]
dir_tabla["p Holm (logística)"] = [coefs["p_holm_dep"].get(NOMBRES[v], np.nan) for v in INTERV]
dir_tabla.index = [NOMBRES[v] for v in INTERV]
display(dir_tabla.loc[[NOMBRES[v] for v in CREDITO] + [NOMBRES[v] for v in INTERV if v not in CREDITO]].round(3))

fig, ax = plt.subplots(figsize=(10.5, 6.5))
ax.grid(False)
M = rho_dir.loc[CREDITO + [v for v in INTERV if v not in CREDITO]]
im = ax.imshow(M.values, cmap="RdBu_r", vmin=-1, vmax=1, aspect="auto")
ax.set_xticks(range(len(NOMS)), NOMS, fontsize=9); ax.set_yticks(range(len(M)), [NOMBRES[v] for v in M.index])
for (i, j), v in np.ndenumerate(M.values):
    ax.text(j, i, f"{v:+.2f}", ha="center", va="center", fontsize=9)
ax.axhline(len(CREDITO) - 0.5, color="black", lw=1)
ax.set_title(f"Correlación (Spearman) entre cada variable y el riesgo predicho en {ANIO_PRUEBA}\nrojo = más valor, más riesgo · azul = más valor, menos riesgo", fontsize=10)
plt.colorbar(im, ax=ax, shrink=0.7)
plt.tight_layout()
plt.show()

mismo_signo = [NOMBRES[v] for v in CREDITO if len({np.sign(rho_dir.loc[v, n]) for n in NOMS}) == 1]
sube = [NOMBRES[v] for v in CREDITO if all(rho_dir.loc[v, n] > 0.1 for n in NOMS)]
baja = [NOMBRES[v] for v in CREDITO if all(rho_dir.loc[v, n] < -0.1 for n in NOMS)]
print("(Las tres columnas se parecen mucho porque las tres probabilidades están muy correlacionadas entre sí, sección 5: la correlación marginal casi no distingue entre modelos; lo que distingue es el odds ratio controlado.)")
print(f"Líneas de crédito en las que los tres modelos coinciden en el signo: {len(mismo_signo)} de {len(CREDITO)}.")
print("Las tres señalan MÁS riesgo cuando sube:", "; ".join(sube) or "ninguna")
print("Las tres señalan MENOS riesgo cuando sube:", "; ".join(baja) or "ninguna")
print("Esto es asociación: el crédito puede estar llegando a municipios que de entrada tienen menos presión sobre el bosque.")
discrepa = [NOMBRES[v] for v in INTERV if NOMBRES[v] in coefs.index and abs(rho_dir.loc[v, "Regresión logística"]) > 0.1
            and np.sign(np.log(coefs.loc[NOMBRES[v], "OR"])) != np.sign(rho_dir.loc[v, "Regresión logística"])]
print("Variables cuyo signo marginal (correlación con el riesgo) es CONTRARIO al de su odds ratio controlado: " + ("; ".join(discrepa) or "ninguna")
      + ". Cuando pasa, la correlación simple refleja que esa variable se mueve junto con el tamaño y el bosque del municipio (la variable estructural), "
        "y el coeficiente controlado dice otra cosa: no debe leerse la correlación como efecto.")
```

**Lectura: dirección, y por qué no basta la correlación simple (ODM-3).**

Los tres coinciden en el signo en las 6 líneas de crédito (más crédito a palma y a cacao se asocia con más riesgo; más crédito total, ganadero y a café, con menos), pero **eso no es evidencia independiente**: las tres probabilidades están tan correlacionadas entre sí (sección 5) que la correlación marginal casi no distingue entre modelos. Además, tres variables cambian de signo al controlar por las demás
(crédito a palma, bovinos por km² y área cosechada): su correlación simple refleja que se mueven con el tamaño y el bosque del municipio. Con control, solo el **cacao (% del municipio)** resiste Holm en la logística (OR de 1,41, p Holm = 0,039); el crédito ganadero (OR de 0,65, p Holm = 0,108) no. Conclusión prudente: asociación descriptiva, sin afirmar que el crédito protege o causa.

### 7.5 ¿Cuáles resisten una prueba formal? (CDM-5)

CDM-5 pide al menos tres variables financieras o de producción con asociación estadísticamente significativa (p < 0,05), **corregida por comparaciones múltiples** y con errores
robustos por conglomerado. Cada técnica la responde a su manera, y no todas las respuestas valen lo mismo: el valor p de la prueba t de la permutación (el del gradient boosting) solo
recoge el azar del barajado y sale "significativo" casi siempre; el *bootstrap* por departamento con Holm sí refleja que los municipios de un departamento se parecen; y los errores
agrupados por departamento de la regresión logística son la versión clásica. Se cuentan solo variables de **intervención**.

```python
filas = []
for n in NOMS:
    d = PERM[n]
    d_int = d[~d.index.isin(ESTRUCTURALES) & (d.importancia_media > 0)]
    filas.append({"modelo": n, "prueba": "permutación, prueba t (como el gradient boosting; optimista)", "n variables": int((d_int.p_valor < 0.05).sum()),
                  "cuáles": ", ".join(NOMBRES[v] for v in d_int.index[d_int.p_valor < 0.05])})
    if "p_holm" in d.columns:
        filas.append({"modelo": n, "prueba": "permutación, bootstrap por departamento + Holm", "n variables": int((d_int.p_holm < 0.05).sum()),
                      "cuáles": ", ".join(NOMBRES[v] for v in d_int.index[d_int.p_holm < 0.05])})
    else:
        filas.append({"modelo": n, "prueba": "permutación, bootstrap por departamento + Holm", "n variables": np.nan,
                      "cuáles": "no calculada"})
ci = coefs[coefs.tipo == "intervención"]
filas.append({"modelo": "Regresión logística", "prueba": "coeficientes con errores por departamento + Holm (¿deforesta?)",
              "n variables": int((ci.p_holm_dep < 0.05).sum()), "cuáles": ", ".join(ci.index[ci.p_holm_dep < 0.05])})
mult = leer_resultado("regresion_logistica_efectos_multiplicativos_magnitud")
mi = mult[mult.tipo == "intervención"]
filas.append({"modelo": "Regresión logística", "prueba": "regresión Gamma con errores por departamento + Holm (¿cuánto?)",
              "n variables": int((mi.p_holm < 0.05).sum()), "cuáles": ", ".join(mi.index[mi.p_holm < 0.05])})
cdm5 = pd.DataFrame(filas)
display(cdm5)

def n_de(modelo, prueba_inicio):
    f = cdm5[(cdm5.modelo == modelo) & cdm5.prueba.str.startswith(prueba_inicio)]
    return f["n variables"].iloc[0]

n_gb, n_rf, n_lr_clas, n_lr_mag = n_de("Gradient boosting", "permutación, bootstrap"), n_de("Random forest", "permutación, bootstrap"), n_de("Regresión logística", "coeficientes"), n_de("Regresión logística", "regresión Gamma")
cumplen = ([f"Gradient boosting ({int(n_gb)})"] if n_gb >= 3 else []) + ([f"Random forest ({int(n_rf)})"] if n_rf >= 3 else []) + ([f"regresión logística en ¿cuánto? ({int(n_lr_mag)})"] if n_lr_mag >= 3 else [])
print(f"Con la prueba exigente (corregida por comparaciones múltiples), el umbral de CDM-5 (≥ 3 variables de intervención) lo cumplen: {', '.join(cumplen) or 'ninguna técnica'}. "
      f"La regresión logística, en ¿deforesta?, llega a {int(n_lr_clas)}.")
robustas = cdm5[cdm5.prueba.str.contains("Holm") & (cdm5["n variables"] > 0)]
variables_robustas = [set(s.split(", ")) for s in robustas["cuáles"] if s and not s.startswith("no calculada")]
comun_robusta = set.intersection(*variables_robustas) if variables_robustas else set()
print(f"Variables que resisten la prueba exigente en TODAS las técnicas que la calcularon: {', '.join(sorted(comun_robusta)) or 'ninguna (cada técnica señala variables distintas)'}.")
```

**Lectura: CDM-5 se cumple solo en parte, y depende de la técnica.**

Con la prueba exigente (bootstrap por departamento y Holm, o errores por departamento en la regresión): bosque aleatorio **3** variables (fincas ganaderas, cacao, crédito ganadero), gradient boosting **2** (fincas ganaderas y crédito ganadero), regresión logística **1** en ¿deforesta? (cacao) y **4** en ¿cuánto?. **Ninguna variable resiste en todas las técnicas.** El cacao está entre las tres primeras de intervención en las tres, pero solo resiste en el
bosque y en la logística; el crédito ganadero resiste en los dos árboles y en la permutación de la logística, pero no en su inferencia. En ¿deforesta? el criterio de tres variables lo cumple solo el bosque aleatorio. En ¿cuánto?, los 4 de la logística son asociaciones dentro de la muestra de un modelo con R² de 0,013, no capacidad de predecir. Para el documento: declarar CDM-5 como cumplido por una técnica
y no por convergencia entre técnicas.

## 8. ¿Dónde coinciden el alto crédito y el alto riesgo? (OBJ-N1, ODM-2)

OBJ-N1 pide identificar al menos un municipio o cadena donde coincidan **alto financiamiento agropecuario y alto riesgo de deforestación**. Se cruza, para cada modelo, el cuarto superior del
riesgo predicho en 2025 con el cuarto superior de crédito de producción medido de dos formas que cuentan cosas distintas: **volumen** (millones de pesos acumulados 2021-2025: ¿cuánta plata llega?) e
**intensidad** (millones por km²: ¿qué tan densa es?). Los municipios grandes de la frontera amazónica reciben mucho en volumen y casi nada por km².

```python
anual_gb = ns_gb["anual"].assign(cod_dane=lambda d: d.cod_dane.astype(str))
vol = anual_gb[anual_gb.anio.between(ANIO_INICIO, ANIO_PRUEBA)].groupby("cod_dane", observed=True).credito_produccion_mm.sum()
muestra_gb = ns_gb["muestra"].assign(cod_dane=lambda d: d.cod_dane.astype(str))
ha_tot = muestra_gb.groupby("cod_dane", observed=True).def_ideam_ha.sum()
intens = muestra_gb.groupby("cod_dane", observed=True).cred.mean()
M = P[["cod_dane", "nombre_municipio", "nombre_departamento"]].assign(vol=P.cod_dane.map(vol).fillna(0), intens=P.cod_dane.map(intens).fillna(0),
                                                                     ha=P.cod_dane.map(ha_tot).fillna(0))
for n in NOMS:
    M[n] = P[f"prob|{n}"].values

cruce = []
sets_volumen = {}
for n in NOMS:
    alto_riesgo = M[n] >= M[n].quantile(0.75)
    for nombre, col in [("volumen", "vol"), ("intensidad", "intens")]:
        alto_cred = M[col] >= M[col].quantile(0.75)
        sel = M[alto_riesgo & alto_cred]
        cruce.append({"modelo": n, "crédito": nombre, "municipios": len(sel), f"% de las hectáreas {ANIO_INICIO}-{ANIO_PRUEBA}": 100 * sel.ha.sum() / M.ha.sum()})
        if nombre == "volumen":
            sets_volumen[n] = set(sel.cod_dane)
cruce = pd.DataFrame(cruce).pivot(index="crédito", columns="modelo", values=["municipios", f"% de las hectáreas {ANIO_INICIO}-{ANIO_PRUEBA}"])
display(cruce.round(1))

en_tres = set.intersection(*sets_volumen.values())
en_alguno_v = set.union(*sets_volumen.values())
tabla_coinciden = (M[M.cod_dane.isin(en_tres)].sort_values("ha", ascending=False)
                   [["nombre_municipio", "nombre_departamento", "vol", "intens", "ha"]]
                   .rename(columns={"vol": f"crédito {ANIO_INICIO}-{ANIO_PRUEBA} (M$)", "intens": "crédito (M$/km²)", "ha": f"ha deforestadas {ANIO_INICIO}-{ANIO_PRUEBA}"}))
print(f"Municipios con alto riesgo Y alto volumen de crédito según los TRES modelos: {len(en_tres)} (según al menos uno: {len(en_alguno_v)}); "
      f"concentran {100 * M[M.cod_dane.isin(en_tres)].ha.sum() / M.ha.sum():.0f} % de las hectáreas deforestadas {ANIO_INICIO}-{ANIO_PRUEBA}.")
display(tabla_coinciden.head(10).round(1))
inten_n = cruce.loc["intensidad", "municipios"]
print(f"Por intensidad (M$/km²) la coincidencia casi desaparece: {', '.join(f'{n} {int(inten_n[n])}' for n in NOMS)} municipios. "
      f"La deforestación está donde el crédito por km² es bajo, aunque el volumen total sea alto.")
```

**Lectura: dónde coinciden alto crédito y alto riesgo (OBJ-N1, ODM-2).**

- **Por volumen sí hay coincidencia:** 42 municipios son de alto riesgo y alto volumen de crédito según los tres modelos (66 según al menos uno) y concentran el 55 % de las hectáreas deforestadas de 2021-2025. Encabezan Cartagena del Chairá, San Vicente del Caguán, Mapiripán, La Macarena y San José del Guaviare.
- **Por intensidad casi no:** solo 1 o 2 municipios (0,2 % a 0,4 % de las hectáreas). Cumaribo recibe 58.000 millones de pesos en total, pero 0,2 millones por km².
- **Cómo leerlo.** El cruce por volumen en buena parte refleja **tamaño**: los municipios grandes reciben más plata en total. La lectura más exigente es la de intensidad: la deforestación está donde llega poco crédito por km², aunque el total sea alto. Con esto OBJ-N1 se cumple en su lectura por volumen, con esa advertencia.

## 9. Ventajas y desventajas

Primero lo que es propio de cada técnica (no depende de una corrida) y, justo después, lo que dicen **los números de esta corrida**, calculado.

**Gradient boosting (LightGBM)** — ensamble de referencia del equipo. Captura interacciones y relaciones no lineales sin suponer una forma funcional; una predicción puntual se explica con SHAP; trae una regresión condicional
para estimar *cuánto*. Es menos directo de explicar a un no especialista que un coeficiente. Su notebook incluye la prueba formal de significancia (*bootstrap* por departamento con Holm) y la comparación estructural vs. intervención (especificaciones A a D), igual que los otros dos.

**Random forest** — promedia muchos árboles, así que es estable, y trae estimación *out-of-bag* de la brecha de sobreajuste. Su notebook incluye bloques y actividades, dependencia parcial y especificaciones con persistencia y región. Pesa más (archivo del modelo y cómputo) y, como el gradient boosting, no entrega un signo por variable sin SHAP o dependencia parcial.

**Regresión logística** — la única que entrega para cada variable un coeficiente con dirección, *odds ratio*, intervalo y valor p con errores robustos por departamento: la cifra que permite justificar *por qué* una variable importa sin apelar a "porque el modelo
lo dice". Es reproducible exactamente. A cambio, supone linealidad en el logit, necesita transformar (log) y podar por multicolinealidad (VIF) —por eso deja variables fuera del modelo—, y su regresión Gamma de "¿cuánto?" es una aproximación.

## 10. Veredicto

```python
def lideres(serie, tol):
    mejor = serie.idxmax()
    empatados = [n for n in serie.index if n != mejor and serie.max() - serie[n] <= tol]
    return mejor, empatados


auc25_s, aucdf_s = cls25.loc["AUC"], clsdf.loc["AUC"]
ba25_s, badf_s = cls25.loc["exactitud balanceada"], clsdf.loc["exactitud balanceada"]
ap25_s = cls25.loc["AP"]
c10 = cobertura.loc["10 %", "hectáreas deforestadas cubiertas (%)"]
m_auc25, e_auc25 = lideres(auc25_s, 0.01)
m_aucdf, e_aucdf = lideres(aucdf_s, 0.01)
m_ba, e_ba = lideres(ba25_s, 0.01)

def frase(medida, serie, mejor, empatados, tol="0,01"):
    s = f"{medida}: {mejor} ({serie[mejor]:.3f})"
    if empatados:
        s += f"; empate técnico (diferencia ≤ {tol}) con " + ", ".join(f"{e} ({serie[e]:.3f})" for e in empatados)
    return s + "."

criterios = pd.DataFrame({f"AUC {ANIO_PRUEBA}": auc25_s, "AUC departamentos fuera": aucdf_s, f"AP {ANIO_PRUEBA}": ap25_s, f"exactitud balanceada {ANIO_PRUEBA}": ba25_s,
                          "exactitud balanceada dep. fuera": badf_s, "hectáreas cubiertas (10 %)": c10, "calibración (Brier)": -brier})
puestos = criterios.rank(ascending=False, method="min").astype(int)
puestos["puesto medio"] = puestos.mean(axis=1)
print("PARA PRIORIZAR MUNICIPIOS POR RIESGO (¿deforesta?)")
print("  " + frase(f"AUC en la prueba temporal {ANIO_PRUEBA}", auc25_s, m_auc25, e_auc25))
print("  " + frase("AUC con departamentos fuera", aucdf_s, m_aucdf, e_aucdf))
print("  " + frase(f"Exactitud balanceada {ANIO_PRUEBA}", ba25_s, m_ba, e_ba))
print(f"  Hectáreas cubiertas revisando el 10 % más riesgoso: {c10.idxmax()} ({c10.max():.0f} %), frente a {c10.idxmin()} ({c10.min():.0f} %).")
print("  Puesto de cada modelo en cada criterio (1 = mejor):")
display(puestos.sort_values("puesto medio"))
mejor_medio = puestos["puesto medio"].idxmin()
cerca = [n for n in NOMS if n != mejor_medio and (auc25_s[mejor_medio] - auc25_s[n]) <= 0.01 and (aucdf_s.max() - aucdf_s[n]) <= 0.01]
print(f"  → Para priorizar: {mejor_medio} (puesto medio {puestos.loc[mejor_medio, 'puesto medio']:.1f})"
      + (f"; {', '.join(cerca)} es una alternativa en empate técnico por AUC (≤ 0,01), así que la elección puede hacerse por interpretabilidad o costo." if cerca else "; los otros quedan a más de 0,01 de AUC."))

print("\nPARA ESTIMAR CUÁNTO (regresión condicional)")
print(f"  Mejor R²: {mejor_r2} ({REG.loc['R²', mejor_r2]:.2f}); mejor ordenando las tasas (Spearman): {mejor_sp} ({REG.loc['Spearman', mejor_sp]:.2f}). "
      f"Ninguno es confiable en valor (R² ≤ {REG.loc['R²'].max():.2f}); sirve, a lo sumo, para ordenar municipios por severidad esperada.")

print("\nPARA JUSTIFICAR UNA DECISIÓN (debida diligencia, explicar el porqué)")
print(f"  La regresión logística: entrega odds ratio, intervalo y valor p corregido; su AUC {ANIO_PRUEBA} es {auc25_s['Regresión logística']:.3f} "
      f"({auc25_s['Regresión logística'] - auc25_s.max():+.3f} frente al mejor), un costo moderado de orden a cambio de poder explicar cada variable.")

print("\nLO QUE LOS TRES DICEN LO MISMO")
print(f"  - Casi todo el poder predictivo es estructural (bosque y área): {', '.join(f'{n} {part_estructural[n]:.0f} %' for n in NOMS)} de la importancia.")
print(f"  - Actividad con mayor peso: {'la misma en los tres (' + top_act.iloc[0] + ')' if acuerdo_act else 'no coincide: ' + '; '.join(f'{n} {a}' for n, a in top_act.items())}.")
print(f"  - Las magnitudes ('¿cuánto?') son débiles en los tres, y el error lo domina Pedraza.")
```

**Lectura del veredicto: la elección depende del propósito, no de diferencias pequeñas.**

El puesto medio pone al gradient boosting primero (1,4), al bosque segundo (2,1) y a la regresión logística tercera (2,4), pero ese ranking es **frágil**: en varios criterios la diferencia es de 0,01 o menos (AUC con departamentos fuera; exactitud balanceada con departamentos fuera, 0,825 para el gradient boosting y la regresión logística) o depende del desempate entre probabilidades saturadas (hectáreas del 10 %).
Lo que los datos sostienen es esto:

- **Para priorizar municipios, los tres son equivalentes** cuando el territorio es nuevo (AUC de 0,905 a 0,911) y el gradient boosting y el bosque lo son además en 2025. Con el 20-30 % de mayor riesgo se cubren las mismas hectáreas. No se puede argumentar que uno "detecta más": con la misma regla de marcado la diferencia desaparece.
- **Para estimar cuánto**, los árboles ordenan mejor (Spearman de 0,72 y 0,73 frente a 0,49), pero ninguno estima hectáreas.
- **Para justificar una decisión**, la regresión logística es la única con coeficiente, intervalo y valor p, a cambio de dejar variables fuera por VIF y de una probabilidad peor calibrada (Brier de 0,131).
- **Diferencias reales entre técnicas:** el orden de "¿cuánto?", la calibración, y qué variables resisten la prueba formal. En lo demás, la observación del tutor de que los modelos "fueron iguales" es correcta.
